```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
import pandas as pd

from nltk.corpus import stopwords
import nltk
import unicodedata

nltk.download("stopwords")

spanish_stopwords = stopwords.words("spanish")

pd.set_option("display.max_rows", 600)
from pathlib import Path
import glob

import altair as alt

alt.data_transformers.enable("vegafusion")
import numpy as np

import operator, joblib
```

```python
directory_path = "../data/lemmas_individual/"
```

```python
text_files = glob.glob(f"{directory_path}/*.txt")
```

```python
text_files
```

```python
text_titles = [Path(text).stem for text in text_files]
```

```python
text_titles
```

```python
tfidf_vectorizer = TfidfVectorizer(
    input="filename",
    stop_words=spanish_stopwords,
    encoding="latin1",
    strip_accents="unicode",
)
```

```python
tfidf_vector = tfidf_vectorizer.fit_transform(text_files)
```

```python
tfidf_vector.shape[0], tfidf_vector.shape[1]
```

```python
terms = list(tfidf_vectorizer.get_feature_names_out())
print("Vocabulary has %d distinct terms" % len(terms))
```

```python
terms
```

```python
def rank_terms(A, term):
    # get the sums over each column
    sums = A.sum(axis=0)
    # map weights to the terms
    weights = {}
    for col, term in enumerate(terms):
        weights[term] = sums[0, col]
    # rank the terms by their weight over all documents
    return sorted(weights.items(), key=operator.itemgetter(1), reverse=True)
```

```python
ranking = rank_terms(tfidf_vector, terms)
for i, pair in enumerate(ranking[0:100]):
    print("%02d. %s (%.2f)" % (i + 1, pair[0], pair[1]))
```

```python
joblib.dump((tfidf_vector, terms, text_titles), "../data/avisos-tfidf.pkl")
```

```python
tfidf_df = pd.DataFrame(
    tfidf_vector.toarray(),  # type: ignore
    index=text_titles,
    columns=tfidf_vectorizer.get_feature_names_out(),
)
```

```python
tfidf_df.loc["00_Document Frequency"] = (tfidf_df > 0).sum()
```

```python
tfidf_slice = tfidf_df[
    [
        "cholito",
        "mestizo",
        "indio",
    ]
]
tfidf_slice.sort_index().round(decimals=2)
```

```python
tfidf_df = tfidf_df.drop("00_Document Frequency", errors="ignore")
```

```python
tfidf_df.stack().reset_index()
```

```python
tfidf_df = tfidf_df.stack().reset_index()
```

```python
tfidf_df = tfidf_df.rename(
    columns={0: "tfidf", "level_0": "document", "level_1": "term", "level_2": "term"}
)
```

```python
tfidf_df.sort_values(by=["document", "tfidf"], ascending=[True, False]).groupby(
    ["document"]
).head(10)
```

```python
top_tfidf = (
    tfidf_df.sort_values(by=["document", "tfidf"], ascending=[True, False])
    .groupby(["document"])
    .head(10)
)
```

```python
top_tfidf.head(40)
```

```python
top_tfidf_plusRand = top_tfidf.copy()
top_tfidf_plusRand["tfidf"] = (
    top_tfidf_plusRand["tfidf"] + np.random.rand(top_tfidf.shape[0]) * 1e-4
)

top_n = 10
top_tfidf_filtered = (
    top_tfidf_plusRand.sort_values(["document", "tfidf"], ascending=[True, False])
    .groupby("document")
    .head(top_n)
)

base = (
    alt.Chart(top_tfidf_filtered)
    .encode(
        x=alt.X(
            "rank:O",
            title="Ranking de términos",
            axis=alt.Axis(labelAngle=0, labelFontSize=10),
        ),
        y=alt.Y("document:N", title="Documento", axis=alt.Axis(labelFontSize=10)),
    )
    .transform_window(
        rank="rank()",
        sort=[alt.SortField("tfidf", order="descending")],
        groupby=["document"],
    )
)

heatmap = base.mark_rect(cornerRadius=4).encode(
    color=alt.Color("tfidf:Q", title="TF-IDF", scale=alt.Scale(scheme="tealblues")),
    tooltip=["term:N", "tfidf:Q", "document:N"],
)

text = base.mark_text(baseline="middle", font="Roboto", fontSize=10).encode(
    text="term:N",
    color=alt.condition("datum.tfidf > 0.25", alt.value("white"), alt.value("black")),
)

chart = (
    (heatmap + text)
    .properties(
        width=700,
        height=alt.Step(22),
        title=alt.TitleParams(
            text="Top términos por documento (TF-IDF)",
            fontSize=18,
            fontWeight="bold",
            anchor="start",
            color="#2a2a2a",
        ),
    )
    .configure_axis(grid=False, labelFont="Roboto", titleFont="Roboto")
    .configure_view(strokeWidth=0)
    .configure_title(font="Roboto")
)

chart
```

```python
top_tfidf_plusRand = top_tfidf.copy()
top_tfidf_plusRand["tfidf"] = (
    top_tfidf_plusRand["tfidf"] + np.random.rand(top_tfidf.shape[0]) * 1e-4
)

top_n_global = 30

global_top = (
    top_tfidf_plusRand.groupby("term")["tfidf"]
    .max()
    .reset_index()
    .sort_values("tfidf", ascending=False)
    .head(top_n_global)
)

chart = (
    alt.Chart(global_top)
    .mark_bar(cornerRadius=3)
    .encode(
        x=alt.X("tfidf:Q", title="TF-IDF"),
        y=alt.Y("term:N", sort="-x", title="Término"),
        color=alt.Color("tfidf:Q", scale=alt.Scale(scheme="tealblues"), legend=None),
        tooltip=["term:N", "tfidf:Q"],
    )
    .properties(
        width=600,
        height=alt.Step(20),
        title=alt.TitleParams(
            text=f"Top {top_n_global} términos con mayor TF-IDF en todos los documentos",
            fontSize=18,
            fontWeight="bold",
            anchor="start",
        ),
    )
    .configure_axis(labelFont="Roboto", titleFont="Roboto")
    .configure_title(font="Roboto")
)

chart
```
